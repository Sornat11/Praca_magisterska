# Uruchomienie Obliczeń i Zestawienie Wyników Eksperymentów
Notatnik ten służy do programowego uruchomienia modeli (ItemKNN, BPR-MF, NeuMF, LightGCN) za pomocą biblioteki RecBole. Wyniki ewaluacji zostaną zebrane do zmiennych i zaprezentowane w tabeli zbiorczej na końcu notatnika.

```python
import pandas as pd
from recbole.quick_start import run_recbole

# Lista na zbiór wyników z poszczególnych modeli
all_results = []
```

## 1. Model: ItemKNN (k-Nearest Neighbors)
Klasyczny model oparty na pamięci (memory-based), wykorzystujący podobieństwo między elementami. Stanowi silny, ale prosty punkt odniesienia (baseline).

```python
print("Uruchamianie ewaluacji dla modelu ItemKNN...")
result_knn = run_recbole(
    model='ItemKNN', 
    dataset='ml-100k', 
    config_file_list=['../2_Experiments/Configs/itemknn.yaml'],
    config_dict={'checkpoint_dir': '../3_Evaluation/Saved/'}
)

knn_metrics = result_knn['test_result']
knn_metrics['Model'] = 'ItemKNN'
all_results.append(knn_metrics)

print(f"Zakończono! NDCG@10: {knn_metrics.get('ndcg@10', 'Brak'):.4f}")
```

## 2. Model: BPR (Bayesian Personalized Ranking - Matrix Factorization)
Tradycyjny model filtrowania kolaboracyjnego bazujący na faktoryzacji macierzy z funkcją straty BPR.

```python
print("Uruchamianie treningu dla modelu BPR...")
result_bpr = run_recbole(
    model='BPR', 
    dataset='ml-100k', 
    config_file_list=['../2_Experiments/Configs/bpr.yaml'],
    config_dict={'checkpoint_dir': '../3_Evaluation/Saved/'}
)

bpr_metrics = result_bpr['test_result']
bpr_metrics['Model'] = 'BPR'
all_results.append(bpr_metrics)

print(f"Zakończono! NDCG@10: {bpr_metrics.get('ndcg@10', 'Brak'):.4f}")
```

## 3. Model: NeuMF (Neural Collaborative Filtering)
Model głębokiego uczenia, łączący Generalized Matrix Factorization (GMF) oraz Multi-Layer Perceptron (MLP).

```python
print("Uruchamianie treningu dla modelu NeuMF...")
result_neumf = run_recbole(
    model='NeuMF', 
    dataset='ml-100k', 
    config_file_list=['../2_Experiments/Configs/ncf.yaml'],
    config_dict={'checkpoint_dir': '../3_Evaluation/Saved/'}
)

neumf_metrics = result_neumf['test_result']
neumf_metrics['Model'] = 'NeuMF'
all_results.append(neumf_metrics)

print(f"Zakończono! NDCG@10: {neumf_metrics.get('ndcg@10', 'Brak'):.4f}")
```

## 4. Model: LightGCN (Graph Neural Network)
Uproszczona grafowa sieć neuronowa do systemów rekomendacyjnych, oparta na splotach po strukturze dwudzielnego grafu (użytkownik-film).

```python
print("Uruchamianie treningu dla modelu LightGCN...")
result_lightgcn = run_recbole(
    model='LightGCN', 
    dataset='ml-100k', 
    config_file_list=['../2_Experiments/Configs/gnn.yaml'],
    config_dict={'checkpoint_dir': '../3_Evaluation/Saved/'}
)

lightgcn_metrics = result_lightgcn['test_result']
lightgcn_metrics['Model'] = 'LightGCN'
all_results.append(lightgcn_metrics)

print(f"Zakończono! NDCG@10: {lightgcn_metrics.get('ndcg@10', 'Brak'):.4f}")
```

## 5. Tabela Zbiorcza Wyników
Zestawienie obliczonych metryk (Recall, MRR, NDCG, Hit) z testowych zbiorów danych dla wszystkich przetrenowanych modeli.

```python
# Budowa i formatowanie tabeli Pandas
df_results = pd.DataFrame(all_results)

# Przeniesienie nazwy modelu na początek tabeli
cols = ['Model'] + [col for col in df_results.columns if col != 'Model']
df_results = df_results[cols]

# Sortowanie wyników od najlepszego do najgorszego na podstawie NDCG@10
df_results = df_results.sort_values(by='ndcg@10', ascending=False).reset_index(drop=True)

# Prezentacja zbiorcza wyników (zaokrąglenie do 4 miejsc po przecinku)
display(df_results.style.format(precision=4).background_gradient(subset=['ndcg@10', 'hit@10'], cmap='Greens'))
```
